## Research Questions

1) Why do we perform hyperparameter tuning after selecting the best baseline models instead of tuning every model from the beginning?
2) What is the difference between model parameters and hyperparameters?
3) What is GridSearchCV? How does it find the best hyperparameter combination?
4) Why is Cross-Validation used inside GridSearchCV? Why don't we simply evaluate each hyperparameter combination on a single train/test split?
5) Suppose GridSearchCV finds a model with a higher Accuracy but a lower Recall than the baseline. Would you automatically choose the tuned model? Why or why not?
6) Imagine your GridSearchCV tries 300 different hyperparameter combinations. Some combinations improve performance by only 0.2% while increasing training time from 10 seconds to 40 minutes. How would you decide whether this improvement is worthwhile?

---

- 1-Hyperparameter tuning is computationally expensive and time-consuming. Therefore, it is more efficient to first evaluate baseline models and identify the most promising candidates. Tuning every model may waste computational resources on algorithms that are fundamentally unsuitable for the problem. By selecting strong baseline models first, we can focus optimization efforts where they are most likely to provide meaningful improvements.

- 2-Model parameters are values learned automatically during training, such as coefficients in Logistic Regression or split rules in Decision Trees. Hyperparameters are configuration values set before training that control the learning process, such as max_depth, n_estimators, or learning_rate. Parameters are learned from data, while hyperparameters are chosen by the practitioner.

- 3-GridSearchCV is a hyperparameter optimization method that systematically evaluates all possible combinations of predefined hyperparameter values. It uses cross-validation to assess each combination and selects the one that achieves the best average validation score according to the chosen evaluation metric.

- 4-Cross-validation provides a more reliable estimate of model performance by evaluating each hyperparameter combination on multiple validation folds. A single train/test split may lead to unstable or biased results because performance can depend heavily on the specific data partition. Cross-validation reduces this risk and helps identify models that generalize better to unseen data.

- 5-No. Model selection should depend on the business objective rather than a single metric. In customer churn prediction, Recall is particularly important because failing to identify customers who are likely to churn can lead to significant business losses. Therefore, I would evaluate the trade-off between Accuracy, Recall, Precision, F1-score, and computational efficiency before selecting the final model.

- 6-I would evaluate whether the performance improvement justifies the additional computational cost. From an engineering perspective, a 0.2% improvement may not be worth a substantial increase in training time and resource consumption. From a business perspective, the selected model should balance predictive performance, efficiency, maintainability, and deployment cost. The goal is not to maximize metrics at any cost, but to choose the most practical and cost-effective solution.

```python
# Imports 
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OrdinalEncoder, StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, classification_report
```

```python
# Custom Functions
def grid_variables(grid_name):
    return {
        'Best Params': grid_name.best_params_,
        'CV Recall': grid_name.best_score_,
    }

def metrics(y_true, y_pred):
    return {
        'Accuracy': accuracy_score(y_true, y_pred),
        'Precision': precision_score(y_true, y_pred), 
        'Test Recall': recall_score(y_true, y_pred), 
        'F1': f1_score(y_true, y_pred), 
    }
```

```python
# Load Clean Dataset
df = pd.read_csv('../data/processed/customer_churn_clean.csv')
df.head()
```

```python
# Feature & Target Separation
X = df.drop(columns='Churn')
y = df['Churn'].map({"No": 0, "Yes": 1})
```

```python
# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(X, y , test_size=0.3, random_state=42, stratify=y)
```

```python
# Feature Lists
binary_features = ['gender', 'Partner', 'Dependents', 'PhoneService', 'PaperlessBilling']

multiclass_features = ['MultipleLines', 'InternetService', 'OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies', 'Contract', 'PaymentMethod']

numerical_features = ['tenure', 'MonthlyCharges', 'TotalCharges']

passthrough_features = [
    "SeniorCitizen"
]
```

```python
# ColumnTransformer
preprocessor = ColumnTransformer(
    transformers=[
        (
            'binary',
            OrdinalEncoder(),
            binary_features
        ),
        (
            'multiclass',
            OneHotEncoder(handle_unknown='ignore'),
            multiclass_features
        ),
        (
            'numerical',
            StandardScaler(),
            numerical_features
        ),
        (
            'pass',
            'passthrough',
            passthrough_features
        )
    ]
)
```

```python
# Building candidated models' pipelines
logistic_pipe = Pipeline([
    ('preprocessor', preprocessor),
    ('logistic_reg', LogisticRegression(random_state=42))
])
rf_pipe = Pipeline([
    ('preprocessor', preprocessor),
    ('rf', RandomForestClassifier(random_state=42))
])
svc_pipe = Pipeline([
    ('preprocessor', preprocessor),
    ('svc', SVC())
])
```

```python
# Logistic ParamGrid
logistic_param_grid = {
    'logistic_reg__C': [0.01, 0.1, 1, 10],
    'logistic_reg__penalty': ['l2'],
    'logistic_reg__solver': ['lbfgs']
}
```

```python
# Building Logistic GridSearchCV
logistic_gridsearch = GridSearchCV(estimator=logistic_pipe, param_grid=logistic_param_grid, cv=5, n_jobs=-1, refit=True, scoring='recall')
```

```python
# Fitting Logistic Grid
logistic_gridsearch.fit(X_train, y_train)
```

```python
# Logistic Important variables
print(grid_variables(logistic_gridsearch))
print(logistic_gridsearch.best_estimator_)
```

```python
# Random Forest ParamGrid
rf_param_grid = {
    "rf__n_estimators": [100, 200, 300],

    "rf__max_depth": [None, 10, 20, 30],

    "rf__min_samples_split": [2, 5, 10],

    "rf__min_samples_leaf": [1, 2, 4],

    "rf__max_features": ["sqrt", "log2"],

    "rf__bootstrap": [True]
}
```

```python
# Building Random Forest GridSearchCV
rf_gridsearch = GridSearchCV(estimator=rf_pipe, param_grid=rf_param_grid, cv=5, n_jobs=-1, refit=True, scoring='recall')
```

```python
# Fitting RF Grid
rf_gridsearch.fit(X_train, y_train)
```

```python
# RF Important variables
print(grid_variables(rf_gridsearch))
print(rf_gridsearch.best_estimator_)
```

```python
# SVM Forest ParamGrid
svc_param_grid = [
    {
        "svc__kernel": ["linear"],
        "svc__C": [0.01, 0.1, 1, 10, 100]
    },

    {
        "svc__kernel": ["rbf"],
        "svc__C": [0.1, 1, 10, 100],
        "svc__gamma": ["scale", "auto", 0.01, 0.001]
    },

    {
        "svc__kernel": ["poly"],
        "svc__degree": [2, 3],
        "svc__C": [0.1, 1, 10],
        "svc__gamma": ["scale", "auto"]
    }
]
```

```python
# Building SVM GridSearchCV
svc_gridsearch = GridSearchCV(estimator=svc_pipe, param_grid=svc_param_grid, cv=5, n_jobs=-1, refit=True, scoring='recall')
```

```python
# Fitting RF Grid
svc_gridsearch.fit(X_train, y_train)
```

```python
# SVM Important variables
print(grid_variables(svc_gridsearch))
print(svc_gridsearch.best_estimator_)
```

```python
# Final models
grid_tuned_logistic = logistic_gridsearch.best_estimator_
grid_tuned_rf = rf_gridsearch.best_estimator_
grid_tuned_svc = svc_gridsearch.best_estimator_
```

```python
# Prediction
tuned_log_pred = grid_tuned_logistic.predict(X_test)
tuned_rf_pred = grid_tuned_rf.predict(X_test)
tuned_svc_pred = grid_tuned_svc.predict(X_test)
```

```python
# Metrics
preds = [tuned_log_pred, tuned_rf_pred, tuned_svc_pred]
model_names = ['LogisticRegression(Tuned)', 'RandomForest(Tuned)', 'SVC(Tuned)']
metrics_list = []
cm_list = []
cr_list = []
for pred in preds:
    metrics_list.append(metrics(y_test, pred))
    cm_list.append(confusion_matrix(y_test, pred))
    cr_list.append(classification_report(y_test, pred))
```

```python
# Confusion Matrix
for model, cm in zip(model_names, cm_list):
    print(f'{model} Confusion Matrix:\n {cm}')
    print(60*'=')
```

```python
# Classification Report
for model, cr in zip(model_names, cr_list):
    print(f'{model} Classification Report:\n {cr}')
    print(60*'=')
```

### Engineering Analysis

The tuning results indicate that model complexity was not the primary limiting factor. Instead, the current feature representation and decision threshold are likely to have a greater impact on further improving churn detection performance.

## Key Findings

- GridSearchCV was successfully applied to the three selected candidate models: Logistic Regression, Random Forest, and Support Vector Machine.

- Logistic Regression retained its original baseline hyperparameters, indicating that the baseline configuration was already optimal within the explored search space.

- Random Forest achieved improved generalization by limiting tree depth (`max_depth=10`) and increasing the minimum number of samples required for splitting (`min_samples_split=10`). However, these changes did not outperform the Logistic Regression baseline.

- For Support Vector Machine, the polynomial kernel produced the highest cross-validation Recall among the evaluated kernel configurations, outperforming the tested linear and RBF alternatives within the defined search space.

- Despite hyperparameter optimization, Logistic Regression continued to provide the best overall balance between Accuracy, Precision, Recall, and F1-score.

- The tuning results suggest that model complexity is not currently the primary limitation. Future improvements are more likely to come from feature engineering, threshold optimization, feature selection, or handling class imbalance rather than further increasing model complexity.
