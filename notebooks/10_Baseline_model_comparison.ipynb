- 1-Comparing multiple baseline models helps identify the most promising algorithms before investing computational resources in hyperparameter tuning. It also provides a fair reference for understanding each model's strengths, weaknesses, and suitability for the business problem.

- 2-Hyperparameter tuning can be computationally expensive, especially for complex models or large search spaces. Therefore, it is generally more efficient to tune only the models that demonstrate strong baseline performance.

- 3-

    - Recall
    - Precision

    - Computational Cost

- 4-Recall is particularly important because it measures how many actual churn customers are correctly identified. Missing a customer who is likely to churn may result in lost revenue and missed retention opportunities.

- 5-Therefore, the final model should maximize business value rather than optimizing a single evaluation metric.

- 6-Technical performance should always be translated into business impact so that stakeholders can understand why a particular model is recommended.

## Engineering Analysis

*Logistic Regression achieved the best overall trade-off between Accuracy, Precision, Recall, and F1-score, making it the strongest baseline model.*

Gaussian Naive Bayes produced the highest Recall, indicating its ability to identify most churn customers. However, this came at the cost of a substantial decrease in Accuracy and Precision, resulting in many false positives.

*Support Vector Machine delivered competitive performance but did not outperform Logistic Regression in terms of Recall or overall business value.*

Random Forest showed moderate performance in its default configuration. Since tree-based ensemble methods often benefit significantly from hyperparameter tuning, it remains a strong candidate for further optimization.

*Decision Tree and KNN did not demonstrate sufficient baseline performance to justify extensive tuning at this stage.*
