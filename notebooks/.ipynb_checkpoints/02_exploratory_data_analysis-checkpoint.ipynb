```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../data/raw/WA_Fn-UseC_-Telco-Customer-Churn.csv')
```

```python
# Target distribution
df['Churn'].value_counts()
```

```python
df['Churn'].value_counts(normalize=True) * 100
```

```python
# Plotting Target Distribution
plt.figure(figsize=(10, 5))
df['Churn'].value_counts().plot(kind='bar')
plt.title("Customer Churn Distribution")
plt.xlabel("Churn")
plt.ylabel("Number of Customers")
plt.show()
```

 The majority of customers remained with the company, while a smaller proportion churned. This indicates that the dataset is moderately imbalanced and evaluation metrics beyond accuracy should be considered.

```python
# Numerical Features Analysis
numeric_columns = df.select_dtypes([np.number]).columns.values
fig, ax = plt.subplots(len(numeric_columns), 2, figsize=(12, 15))
for i, feature in enumerate(numeric_columns):
    ax[i, 0].hist(df[feature])
    ax[i, 0].set_title(f'{feature} Distribution Histogram')
    ax[i, 0].set_xlabel('Value')
    ax[i, 0].set_ylabel('Frequency')
    df.boxplot(
        column=feature,
        by='Churn',
        ax=ax[i, 1]
    )
    ax[i, 1].set_title(f'{feature} Distribution Boxplot')
    ax[i, 1].set_ylabel('Distribution')
plt.tight_layout()
plt.show()
```

```python
sns.countplot(x=df["SeniorCitizen"])
sns.countplot(x=df["SeniorCitizen"], hue=df["Churn"])
```

## SeniorCitizen
### Observation

Although `SeniorCitizen` is stored as an integer, it represents a binary categorical feature rather than a continuous numerical variable.

Most customers are not senior citizens, indicating an imbalanced distribution.

### Business Insight

Since this variable is categorical by nature, it should be analyzed using categorical visualizations such as count plots instead of numerical distribution plots.

## tenure
### Observation

The distribution of customer tenure is not normally distributed. A large number of customers are either relatively new or have remained with the company for a long period.

The boxplot shows that customers who churned generally have much shorter tenure than those who stayed.

### Business Insight

Customer loyalty appears to increase with longer tenure. Newer customers are more likely to leave the company, suggesting that early customer engagement and retention strategies may significantly reduce churn.

## MonthlyCharges
### Observation

Monthly charges show a non-normal and multi-modal distribution, indicating the existence of different pricing plans or customer segments.

Customers who churned tend to have higher monthly charges compared to those who remained.

### Business Insight

Higher monthly costs may be associated with increased churn risk. Customers paying higher fees may have greater expectations regarding service quality, making them more sensitive to dissatisfaction.

```python
# Categorical Features Analysis
categorical_columns = (
    df.select_dtypes(include="object")
      .columns
      .drop(["customerID", "TotalCharges", "Churn"])
)
for feature in categorical_columns:
    print(pd.crosstab(df[feature], df["Churn"], normalize='index') * 100)
    print('='*60)
```

```python
fig, ax = plt.subplots(5, 3, figsize=(18, 24))
ax = ax.flatten()
for i, feature in enumerate(categorical_columns):
    sns.countplot(
        data=df,
        x=feature,
        hue="Churn",
        ax=ax[i]
    )
    ax[i].set_title(f"Customer Churn by {feature} Type")
    ax[i].tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()
```

## Gender

### Observation

The distribution of churn appears to be very similar for both male and female customers. No noticeable difference in churn behavior is observed between the two groups.

### Business Insight

Gender does not appear to be a strong factor influencing customer churn and is unlikely to be a primary driver of customer retention.

## Partner

### Observation

Customers without a partner appear to churn more frequently than those living with a partner.

### Business Insight

Customers living alone may have different service needs or lower long-term commitment. Retention campaigns could be tailored to this customer segment.

## Dependents

### Observation

Customers without dependents show a noticeably higher churn rate compared to customers who have dependents.

### Business Insight

Customers with dependents may rely more heavily on the service, making them more likely to remain loyal.

## PhoneService

### Observation

Most customers subscribe to the phone service. The difference in churn between customers with and without phone service is not substantial based on the observed counts alone.

### Business Insight

Phone service alone does not appear to provide strong evidence for predicting customer churn.

## MultipleLines

### Observation

Customers with and without multiple phone lines exhibit relatively similar churn behavior.

### Business Insight

The number of phone lines does not appear to have a strong relationship with customer churn.

## InternetService ⭐

### Observation

Customers using fiber optic internet appear to experience considerably higher churn rates than DSL customers, while customers without internet service show the lowest churn.

### Business Insight

Fiber optic customers may have higher expectations regarding service quality or pricing. This customer segment deserves further investigation to identify potential causes of dissatisfaction.

## OnlineSecurity ⭐

### Observation

Customers without the Online Security service exhibit substantially higher churn compared to customers who subscribe to this service.

### Business Insight

Providing additional security-related services may improve customer retention and increase perceived service value.

## OnlineBackup

### Observation

Customers who do not subscribe to Online Backup appear to churn more frequently than those who use this service.

### Business Insight

Value-added services such as Online Backup may contribute to stronger customer engagement and long-term retention.

## DeviceProtection

### Observation

Customers without Device Protection show a higher tendency to churn than customers who subscribe to the service.

### Business Insight

Additional protection services may increase customer commitment and reduce the likelihood of churn.

## TechSupport ⭐⭐⭐

### Observation

A clear difference is observed between customers with and without Technical Support. Customers lacking technical support appear to churn much more frequently.

### Business Insight

Providing accessible and effective technical support could play an important role in reducing customer churn and improving customer satisfaction.

## StreamingTV

### Observation

The difference in churn between customers with and without Streaming TV is relatively small.

### Business Insight

Streaming TV alone does not appear to be a major factor influencing customer churn.

## StreamingMovies

### Observation

Similar to Streaming TV, only a minor difference in churn behavior is observed between customer groups.

### Business Insight

Movie streaming services alone are unlikely to be strong predictors of customer churn.

## Contract ⭐⭐⭐⭐⭐

### Observation

Contract type shows one of the strongest relationships with customer churn. Customers on month-to-month contracts exhibit significantly higher churn rates, whereas customers with one-year and two-year contracts are considerably more likely to remain with the company.

### Business Insight

Encouraging customers to switch from month-to-month plans to long-term contracts could substantially improve customer retention. Loyalty incentives and contract renewal discounts may help reduce churn.

## PaperlessBilling

### Observation

Customers using paperless billing appear to churn more frequently than those receiving traditional paper bills.

### Business Insight

This observation may reflect differences in customer preferences or demographics rather than the billing method itself. Further investigation is recommended before drawing conclusions.

## PaymentMethod ⭐

### Observation

Customers using electronic checks appear to have noticeably higher churn rates than customers using other payment methods.

### Business Insight

The electronic check customer segment may require additional investigation to determine whether payment experience, customer profile, or other factors contribute to the higher churn rate.

### Data Quality Issue: TotalCharges

The `TotalCharges` feature was initially loaded as an object rather than a numeric column.

Further investigation revealed that the column contains a small number of blank string values (" ") instead of valid numerical values.

These blank entries most likely correspond to newly joined customers who have not yet accumulated any total charges.

This issue will be addressed during the data cleaning stage by replacing blank strings with missing values and converting the column to a numeric data type.

## Key Findings

The exploratory analysis revealed several important patterns related to customer churn:

- Contract type appears to be one of the strongest indicators of customer churn.
- Customers using fiber optic internet services tend to churn more frequently.
- Customers without Online Security or Technical Support exhibit noticeably higher churn rates.
- Customers with shorter tenure are more likely to leave the company.
- Higher monthly charges appear to be associated with increased churn.
- Gender and MultipleLines show little evidence of a strong relationship with customer churn.

These findings provide valuable business insights and will guide the preprocessing, feature engineering, and model development stages in subsequent sprints.
